# Experimental WAAM summary
This notebook reproduces the two quantitative figures included in the repository from the curated public summary.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data/processed/summary_metrics.csv')
df
```

```python
ax = df.plot.bar(x='condition', y='average_vickers_hardness_HV10', legend=False)
ax.set_ylabel('Average Vickers hardness (HV10)')
ax.set_xlabel('WAAM wall condition')
plt.tight_layout()
```

```python
ax = df.plot.bar(x='condition', y='average_grain_size_um', legend=False)
ax.set_ylabel('Average grain size (µm)')
ax.set_xlabel('WAAM wall condition')
plt.tight_layout()
```

## Interpretation boundary
These plots describe W1-W3 as combined deposition conditions. They do not isolate heat input from bead strategy or other process variables.
